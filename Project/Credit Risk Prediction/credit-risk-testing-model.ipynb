```python
model_load = pickle.load(open('credit-risk-prediction.pkl', 'rb'))
```

```python
low_risk = pd.DataFrame([{
    'Age': 35,
    'Income': 90000,
    'LoanAmount': 15000,
    'CreditScore': 750,
    'YearsExperience': 10,
    'Gender': 'Female',
    'Education': 'Masters',
    'City': 'New York',
    'EmploymentType': 'Salaried'
}])
```

```python
predict = model_load.predict(low_risk)
prob = model_load.predict_proba(low_risk)

print(predict)
print(prob)
```

```python
status = "Loan Approved" if predict[0] == 1 else "Loan Rejected"
print(f"Decision       : {status}")
print(f"Approval Prob  : {prob[0][1]:.2%}")
print(f"Rejection Prob : {prob[0][0]:.2%}")
```
