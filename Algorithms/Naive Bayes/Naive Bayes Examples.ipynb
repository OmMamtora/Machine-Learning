### For Outlook

```python
# Outlook

pd.crosstab(df['outlook'], df['play'])
```

```python
p_outlook_overcast_no = 0
p_outlook_rain_no = 2/5
p_outlook_sunny_no = 3/5


p_outlook_overcast_yes = 4/9
p_outlook_rain_yes = 3/9
p_outlook_sunny_yes = 2/9
```

### For Temprature

```python
pd.crosstab(df['temp'], df['play'])
```

```python
p_temp_cool_no = 1/5
p_temp_hot_no = 2/5
p_temp_mild_no = 2/5

p_temp_cool_yes = 3/9
p_temp_hot_yes = 2/9
p_temp_mild_yes = 4/9
```

### For Humidity

```python
pd.crosstab(df['humidity'], df['play'])
```

```python
p_humidity_high_no = 4/5
p_humidity_normal_no = 1/5

p_humidity_high_yes = 3/9
p_humidity_normal_yes = 6/9
```

### For Wind

```python
pd.crosstab(df['wind'], df['play'])
```

```python
p_wind_strong_no = 3/5
p_wind_weak_no = 2/5

p_wind_strong_yes = 3/9
p_wind_weak_yes = 6/9
```

# Problem 1:

### Outlook=Sunny, Temp=Hot, Humidity=High, Wind=Weak
#### Play or No Play?

### Solution

#### P(Yes | Sunny,Hot,High,Weak)=P(Sunny | Yes) * P(Hot|Yes) * P(High|Yes) * P(Weak | Yes) * P(Yes)

#### P(No | Sunny,Hot,High,Weak)=P(Sunny |No) * P(Hot|No) * P(High|No) * P(Weak | No) * P(No)

#### Compare and decide using the Maximum a posteriori rule

```python
# Outlook=Sunny, Temp=Hot, Humidity=High, Wind=Weak

p_play_yes = p_outlook_sunny_yes * p_temp_hot_yes * p_humidity_high_yes * p_wind_weak_yes * p_yes
print(p_play_yes)

p_play_no = p_outlook_sunny_no * p_temp_hot_no * p_humidity_high_no * p_wind_weak_no * p_no
print(p_play_no)

if p_play_yes > p_play_no:
    print("Play")
else:
    print("No Play")
```

# Problem 2

### Solution

#### P(No | Overcast,Cold, Low,Weak)=P(Overcast | No) * P(Cold | No) * P(Low | No) * P(Weak | No) * P(No)

#### Compare and decide using the Maximum a posteriori rule

```python
p_play_yes2 = p_outlook_overcast_yes * p_temp_cool_yes * p_humidity_normal_yes * p_wind_weak_yes * p_yes
print(p_play_yes2)

p_play_no2 = p_outlook_overcast_no * p_temp_cool_no * p_humidity_normal_no * p_wind_weak_no * p_no
print(p_play_no2)

if p_play_yes2 > p_play_no2:
    print("Play")
else:
    print("No Play")
```
