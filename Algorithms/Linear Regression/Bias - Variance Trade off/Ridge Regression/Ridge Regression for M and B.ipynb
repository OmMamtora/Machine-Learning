```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge
```

```python
X, y = make_regression(n_samples = 100, n_features = 1, n_informative = 1, n_targets = 1, noise = 20, random_state = 13)
```

```python
plt.scatter(X, y)
```

```python
lr = LinearRegression()
lr.fit(X, y)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
plt.scatter(X, y)
plt.plot(X, lr.predict(X), color = 'red')
```

## Ridge Regression

```python
rr = Ridge(alpha = 10)
rr.fit(X, y)
```

```python
print(rr.coef_)
print(rr.intercept_)
```

```python
rr1 = Ridge(alpha = 100)
rr1.fit(X, y)
```

```python
print(rr1.coef_)
print(rr1.intercept_)
```

```python
# subplot(Row , Column, Position)

plt.figure(figsize=(15,5))
plt.subplot(1,2,1)
plt.scatter(X, y)
plt.plot(X, rr.predict(X), color = 'red')
plt.title(f"Value of alpha = {rr.alpha}")

plt.subplot(1,2,2)
plt.scatter(X, y)
plt.plot(X, rr1.predict(X), color = 'green')
plt.title(f"Value of alpha = {rr1.alpha}")
```

```python
plt.plot(X, y, 'b.')
plt.plot(X, lr.predict(X), color = 'red', label = 'alpha = 0')
plt.plot(X, rr.predict(X), color = 'green', label = 'alpha = 10')
plt.plot(X, rr1.predict(X), color = 'brown', label = 'alpha = 100')
plt.legend()
```
