```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDRegressor, Ridge
from sklearn.metrics import r2_score
```

```python
X, y = load_diabetes(return_X_y = True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 4)
```

## Using SGDRegressor

```python
reg = SGDRegressor(penalty = 'l2', max_iter = 500, eta0 = 0.1, learning_rate = 'constant', alpha = 0.001)
```

```python
reg.fit(X_train, y_train)

y_pred = reg.predict(X_test)
print("R2 Score", r2_score(y_test, y_pred))
print(reg.coef_)
print()
print(reg.intercept_)
```

## Using Ridge Class

```python
re = Ridge(alpha = 0.001, max_iter = 500, solver = 'sparse_cg')
```

```python
re.fit(X_train, y_train)

y_pred = re.predict(X_test)
print("R2 Score", r2_score(y_test, y_pred))
print(re.coef_)
print()
print(re.intercept_)
```

```python
class RidgeGD:
    def __init__(self, epochs, learning_rate , alpha):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.alpha = alpha
        self.coef = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        self.coef_ = np.ones(X_train.shape[1])
        self.intercept_ = 0
        thetha = np.insert(self.coef_, 0, self.intercept_) # thetha = w

        X_train = np.insert(X_train, 0, 1, axis = 1)

        for i in range(self.epochs):
            thetha_derivative = np.dot(X_train.T, X_train).dot(thetha) - np.dot(X_train.T, y_train) - self.alpha * thetha
            thetha = thetha - self.learning_rate * thetha_derivative

        self.coef_ = thetha[1:]
        self.intercept_ = thetha[0]


    def predict(self, X_test):
        return (X_test * self.coef_) + self.intercept_
```

```python
rgd = RidgeGD(epochs = 500, alpha = 0.001, learning_rate = 0.005)

```

```python
rgd.fit(X_train, y_train)
print(rgd.coef_)
print(rgd.intercept_)
```
