```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
```

```python
X, y = make_regression(n_samples = 4, n_features = 1, n_informative = 1, n_targets = 1, noise = 80, random_state = 13)
```

```python
plt.scatter(X, y)
```

```python
reg = LinearRegression()

reg.fit(X, y)
```

## coef_ :-> Slope (S)

```python
reg.coef_
```

## intercept_ :-> intercept_ (C)

```python
reg.intercept_
```

```python
plt.scatter(X, y)

plt.plot(X, reg.predict(X), color = 'red')
```

### Let's apply gradient descent assume slope value constant m = 78.35
### and assume it's started with intercept b = 0

```python
y_pred = ((78.35 * X) + 0).reshape(4)
```

```python
plt.scatter(X, y)
plt.plot(X, reg.predict(X), color = 'red', label = 'OLS')
plt.plot(X, y_pred, color = 'green', label = 'b = 0')
plt.legend()
plt.show()
```

## Iteration 1

## how much and in which direction b should change to reduce prediction error.

```python
m = 78.35
b = 0

loss_slope = -2 * np.sum(y - (m * X.ravel() - b))

loss_slope
```

### Here lr = Learning Rate

```python
lr = 0.1

step_size = loss_slope * lr
step_size
```

## Calculate new Intercept

```python
b = b-step_size
b
```

```python
y_pred1 = ((78.35 * X) + b).reshape(4)

plt.scatter(X, y)
plt.plot(X, reg.predict(X), color = 'red', label = 'OLS')
plt.plot(X, y_pred, color = 'green', label = 'b = 0')
plt.plot(X, y_pred1, color ='blue', label = 'b = {}'.format(b))

plt.legend()
plt.show()
```

## Iteration 2

```python
loss_slope = -2 * np.sum(y - m * X.ravel() - b)

loss_slope
```

```python
step_size = loss_slope * lr
step_size
```

```python
b = b - step_size

b
```

```python
y_pred2= ((78.35 * X) + b).reshape(4)

plt.scatter(X, y)
plt.plot(X, reg.predict(X), color = 'red', label = 'OLS')
plt.plot(X, y_pred, color = 'green', label = 'b = 0')
plt.plot(X, y_pred1, color ='orange', label = 'b = {}'.format(b))
plt.plot(X, y_pred2, color ='blue', label = 'b = {}'.format(b))



plt.legend()
plt.show()
```

## Iteration 3

```python
loss_slope = -2 * np.sum(y - m * X.ravel() - b)

loss_slope
```

```python
step_size = loss_slope * lr

step_size
```

```python
b = b - step_size

b
```

```python
y_pred3= ((78.35 * X) + b).reshape(4)

plt.scatter(X, y)
plt.plot(X, reg.predict(X), color = 'red', label = 'OLS')
plt.plot(X, y_pred, color = 'green', label = 'b = 0')
plt.plot(X, y_pred1, color ='orange', label = 'b = {}'.format(b))
plt.plot(X, y_pred2, color ='blue', label = 'b = {}'.format(b))
plt.plot(X, y_pred3, color ='black', label = 'b = {}'.format(b))


plt.legend()
plt.show()
```

```python
b = -100
m = 78.35
lr = 0.01

epochs = 50

for i in range(epochs):
    loss_slope = -2 * np.sum(y - m * X.ravel() - b)
    b = b - (lr * loss_slope)

    y_pred = m * X + b

    plt.plot(X,y_pred)

plt.scatter(X,y)
```
