```python
import pandas as pd
import numpy as np

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('train.csv')[['Age', 'Pclass', 'SibSp', 'Parch', 'Survived']]
df.head()
```

```python
df.shape
```

```python
df.dropna(inplace = True)
```

```python
df.shape
```

```python
X = df.iloc[:, :4]
y = df.iloc[:, -1]
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), X, y, scoring = 'accuracy', cv = 20)) # cv = 20 means we do cross validation 20 times
```

# Applying Featrure Construction

```python
X['Family_size'] = X['SibSp'] + X['Parch'] + 1 # here we do +1 why because that person itself if a member of family
```

```python
X.head()
```

```python
X[X['Family_size'] > 3]
```

```python
(X['Family_size'] > 3).value_counts()
```

```python
def family_type(num):
    if num ==1:
        return 0 # alone
    elif num > 1 and num <=4:
        return 1 # Samll Family
    else:
        return 2 # Large Family
```

```python
X['Family_type'] = X['Family_size'].apply(family_type)
```

```python
X.head()
```

```python
X.drop(columns = ['SibSp', 'Parch', 'Family_size'], inplace = True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), X, y, scoring = 'accuracy', cv = 20))
```
