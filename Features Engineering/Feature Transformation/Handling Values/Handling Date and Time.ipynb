# Working With Date

```python
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

## 1. Extract Year

### dt :- dt means DateTime

```python
date['date_year'] = date['date'].dt.year
```

```python
date.head()
```

## 2. Extract Month

```python
date['date_month'] = date['date'].dt.month
```

```python
date.head()
```

## 3. Month Name

```python
date['date_month_name'] = date['date'].dt.month_name()
```

```python
date.head()
```

## 4. Extract Date

```python
date['date_day'] = date['date'].dt.day
```

```python
date.head()
```

## 5. Day of Week

```python
date['date_day_of_week'] = date['date'].dt.dayofweek
```

```python
date.head()
```

## 6. Day of week - name

```python
date['date_dow_name'] = date['date'].dt.day_name()
```

```python
date.head()
```

## 7. Is weekend or not?

```python
date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Saturday','Sunday']),1,0)
```

```python
date.head()
```

## 8. Extract Quarter

```python
date['date_quarter'] = date['date'].dt.quarter
```

```python
date.head()
```

## 9. Extract semester

```python
date['date_semester'] = np.where(date['date_quarter'].isin([1,2]), 1, 2)
```

```python
date.head()
```

## 10. Extract Time elapsed between dates

```python
date['date_week_of_year'] = date['date'].dt.week
```

```python
date.
```

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

# Working with time

```python
time['date'] = pd.to_datetime(time['date'])
```

```python
time.info()
```

```python
time['hour'] = time['date'].dt.hour
time['min'] = time['date'].dt.minute
time['second'] = time['date'].dt.second
```

```python
time.head()
```

## Extract time part

```python
time['time'] = time['date'].dt.time
```

```python
time.head()
```

## Extract Time difference

```python
today - time['date']
```

## Time in seconds

```python
(today - time['date']) / np.timedelta64(1, 's')
```

## Time in Minutes

```python
(today - time['date']) / np.timedelta64(1, 'm')
```

## Time in hours

```python
(today - time['date']) / np.timedelta64(1, 'h')
```
