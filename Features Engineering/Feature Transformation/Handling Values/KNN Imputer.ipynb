```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('titanic_train.csv', usecols=['Age','Pclass','Fare','Survived'])
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns = ['Survived']), df['Survived'], test_size = 0.2, random_state = 2)
```

```python
X_train.head()
```

```python
knn = KNNImputer(n_neighbors=1, weights = 'distance')

X_train_trf = knn.fit_transform(X_train)
X_test_trf = knn.transform(X_test)
```

```python
pd.DataFrame(X_train_trf, columns = X_train.columns)
```

```python
lr = LogisticRegression()

lr.fit(X_train_trf, y_train)

y_pred = lr.predict(X_test_trf)

print(accuracy_score(y_pred, y_test))
```

```python
si = SimpleImputer()

X_train_trf2 = si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)
```

```python
lr2 = LogisticRegression()

lr2.fit(X_train_trf2, y_train)

y_pred2 = lr2.predict(X_test_trf2)

print(accuracy_score(y_test,y_pred2))
```
