```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('titanic.csv')
df.head()
```

```python
df['number'].unique()
```

```python
fig = df['number'].value_counts().plot.bar()
fig.set_title('Passanger Travelling with')
```

## Extract Numeric Part

```python
df['number_numeric'] = pd.to_numeric(df['number'], errors = 'coerce', downcast = 'integer')
```

```python
df['number_numeric']
```

## Extract Categorical Part

```python
df['number_categorical'] = np.where(df['number_numeric'].isnull(),df['number'], np.nan)
```

```python
df['number_categorical']
```

```python
df['cabin_num'] = df['Cabin'].str.extract('(\d+)')  # Capture Numeric Part
df['cabin_category'] = df['Cabin'].str[0]  # Capture First letter


df.head()
```

```python
df['cabin_category'].value_counts().plot(kind = 'bar')
```

### Extract the last bit of ticket as number

```python
df['ticket_num'] = df['Ticket'].apply(lambda s : s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'], errors = 'coerce', downcast = 'integer')
```

### Extract the first part of ticker  as categoty

```python
df['ticket_category'] = df['Ticket'].apply(lambda s : s.split()[0])
df['ticket_category'] = np.where(df['ticket_category'].str.isdigit(), np.nan, df['ticket_category'])

df.head(20)
```
